**35. Least Privilege Access Audit**

```python
required_permissions = {
    "Admin": ["read", "write", "delete"],
    "Employee": ["read"],
    "Manager": ["read", "write"]
}

users = {
    "Ravi": ("Employee", ["read", "write"]),
    "John": ("Admin", ["read", "write", "delete"]),
    "Sam": ("Manager", ["read", "write"])
}

for user, (role, permissions) in users.items():

    excess = set(permissions) - set(required_permissions[role])

    if excess:
        print(user, "- Excess Access:", excess)
    else:
        print(user, "- Access is Correct")
```

**37. Continuous Security Monitoring**

```python
baseline = 2

login_events = [
    ("admin", "success"),
    ("john", "failed"),
    ("john", "failed"),
    ("john", "failed"),
    ("admin", "success")
]

failed_count = {}

print("Monitoring Login Events...\n")

for user, status in login_events:

    if status == "failed":
        failed_count[user] = failed_count.get(user, 0) + 1

    print(user, "-", status)

    if failed_count.get(user, 0) > baseline:
        print("ALERT: Unusual Login Activity!")
```

**38. Risk Score Calculation**

```python
likelihood = int(input("Enter likelihood (1-5): "))
impact = int(input("Enter impact (1-5): "))

score = likelihood * impact

print("Risk Score:", score)

if score <= 4:
    print("Risk Level: Low")

elif score <= 9:
    print("Risk Level: Medium")

elif score <= 16:
    print("Risk Level: High")

else:
    print("Risk Level: Critical")
```

**39. Threat Intelligence Platform (TIP) Simulation**

```python
# Threat intelligence feed
ioc_feed = ["192.168.1.100", "malicious.com"]

# Normalize indicators
normalized_iocs = [ioc.lower() for ioc in ioc_feed]

# Internal logs
logs = [
    "User connected to 192.168.1.100",
    "User visited google.com",
    "Connection to malicious.com"
]

print("Threat Intelligence Alerts:")

for log in logs:
    for ioc in normalized_iocs:
        if ioc in log.lower():
            print("ALERT: Malicious IOC Found:", ioc)
```
