**10. Simulate a Stateful Inspection Firewall**

```python
# Store established outbound connections
connection_table = set()

def firewall(packet):
    src = packet["src"]
    dst = packet["dst"]
    direction = packet["direction"]
    protocol = packet["protocol"]
    sport = packet["sport"]
    dport = packet["dport"]

    if direction == "OUTBOUND":
        connection = (src, sport, dst, dport, protocol)
        connection_table.add(connection)
        return "ALLOWED: Outbound connection established"

    # Match inbound traffic against a tracked connection
    reverse = (dst, dport, src, sport, protocol)

    if reverse in connection_table:
        return "ALLOWED: Return traffic for established connection"

    return "BLOCKED: Unsolicited inbound traffic"


# Simulated packets
packets = [
    {
        "src": "192.168.1.10", "dst": "203.0.113.20",
        "direction": "OUTBOUND", "protocol": "TCP",
        "sport": 50000, "dport": 443
    },
    {
        "src": "203.0.113.20", "dst": "192.168.1.10",
        "direction": "INBOUND", "protocol": "TCP",
        "sport": 443, "dport": 50000
    },
    {
        "src": "198.51.100.5", "dst": "192.168.1.10",
        "direction": "INBOUND", "protocol": "TCP",
        "sport": 4444, "dport": 22
    }
]

for packet in packets:
    print(firewall(packet))
```

**13. Assess Whether Data Qualifies as Threat Intelligence**

```python
def assess_intelligence(context, confidence, actionable):
    score = 0

    if context.strip():
        score += 1

    if confidence >= 70:
        score += 1

    if actionable.strip():
        score += 1

    print("\n--- Threat Intelligence Assessment ---")
    print("Context:", context or "Missing")
    print("Confidence:", str(confidence) + "%")
    print("Actionability:", actionable or "Missing")
    print("Assessment Score:", str(score) + "/3")

    if score == 3:
        print("Classification: Actionable Threat Intelligence")
    elif score >= 1:
        print("Classification: Partially Enriched Data")
    else:
        print("Classification: Raw Data")


context = input("Enter threat context: ")

try:
    confidence = int(input("Enter confidence (0-100): "))

    if not 0 <= confidence <= 100:
        print("Confidence must be between 0 and 100.")
    else:
        actionable = input(
            "Enter recommended action (or leave blank): "
        )
        assess_intelligence(context, confidence, actionable)

except ValueError:
    print("Please enter a valid confidence score.")
```

**14. SIEM Architecture Pipeline: Collection, Normalization, Correlation, and Alerting**

```python
from collections import defaultdict
from datetime import datetime

# 1. Collect raw events from multiple sources
raw_events = [
    {
        "source": "Firewall",
        "time": "10:00",
        "ip": "192.0.2.15",
        "event": "failed_login"
    },
    {
        "source": "Authentication",
        "time": "10:01",
        "ip": "192.0.2.15",
        "event": "failed_login"
    },
    {
        "source": "Authentication",
        "time": "10:02",
        "ip": "192.0.2.15",
        "event": "failed_login"
    },
    {
        "source": "Endpoint",
        "time": "10:03",
        "ip": "192.0.2.20",
        "event": "file_access"
    }
]

print("1. Raw Events Collected:", len(raw_events))

# 2. Normalize events into a common format
normalized_events = []

for event in raw_events:
    normalized = {
        "timestamp": event["time"],
        "source": event["source"],
        "source_ip": event["ip"],
        "event_type": event["event"].lower()
    }
    normalized_events.append(normalized)

print("\n2. Normalized Events:")
for event in normalized_events:
    print(event)

# 3. Correlate failed logins by source IP
failed_logins = defaultdict(int)

for event in normalized_events:
    if event["event_type"] == "failed_login":
        failed_logins[event["source_ip"]] += 1

# 4. Generate alerts when threshold is reached
threshold = 3
alerts = []

for ip, count in failed_logins.items():
    if count >= threshold:
        alerts.append({
            "source_ip": ip,
            "failed_attempts": count,
            "severity": "HIGH",
            "alert": "Possible brute-force activity"
        })

print("\n3. Correlation Results:")
print(dict(failed_logins))

print("\n4. SIEM Alerts:")
if alerts:
    for alert in alerts:
        print(alert)
else:
    print("No alerts generated.")
```

**15. Classify a Security Incident by Root Cause**

```python
def classify_incident(narrative):
    text = narrative.lower()

    categories = {
        "Phishing": [
            "phishing email", "malicious link",
            "fake login page", "stolen credentials"
        ],
        "Unpatched Vulnerability": [
            "unpatched", "missing security patch",
            "known vulnerability", "outdated software exploited"
        ],
        "Misconfiguration": [
            "misconfigured", "publicly accessible storage",
            "default password", "incorrect firewall rule"
        ],
        "Insider Threat": [
            "employee stole", "insider",
            "employee leaked", "unauthorized employee access"
        ],
        "Supply-Chain Compromise": [
            "compromised software update",
            "third-party vendor", "malicious dependency",
            "supplier compromise"
        ]
    }

    scores = {
        category: sum(
            keyword in text for keyword in keywords
        )
        for category, keywords in categories.items()
    }

    highest = max(scores.values())

    if highest == 0:
        return "Root cause not identified"

    winners = [
        category for category, score in scores.items()
        if score == highest
    ]

    if len(winners) > 1:
        return "Ambiguous: " + ", ".join(winners)

    return winners[0]


narrative = input("Enter incident case study: ")
print("Likely Root Cause:", classify_incident(narrative))
```

**16. Defense-in-Depth Security Simulation**

```python
defense_layers = [
    ("Firewall", True),
    ("Intrusion Detection System", False),
    ("Endpoint Protection", True),
    ("Multi-Factor Authentication", True),
    ("Access Control", True)
]

def simulate_attack(layers):
    print("Defense-in-Depth Simulation\n")

    for layer, blocks_attack in layers:
        print("Checking:", layer)

        if blocks_attack:
            print("Result: ATTACK BLOCKED")
            print("Remaining layers are not reached.")
            return True
        else:
            print("Result: Attack passes this layer\n")

    print("WARNING: Attack passed all security layers!")
    return False


protected = simulate_attack(defense_layers)

print("\nFinal Status:",
      "System Protected" if protected else "System Compromised")
```
