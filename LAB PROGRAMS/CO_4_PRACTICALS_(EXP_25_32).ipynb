**25. Evidence Chain-of-Custody Verification**

```python
import hashlib

records = [
    {"evidence": "Laptop", "time": 100, "previous_hash": "0"},
    {"evidence": "USB", "time": 110, "previous_hash": ""}
]

# Generate hash for first record
records[0]["hash"] = hashlib.sha256(
    (records[0]["evidence"] + str(records[0]["time"]) +
     records[0]["previous_hash"]).encode()
).hexdigest()

# Link second record to first
records[1]["previous_hash"] = records[0]["hash"]

records[1]["hash"] = hashlib.sha256(
    (records[1]["evidence"] + str(records[1]["time"]) +
     records[1]["previous_hash"]).encode()
).hexdigest()

valid = True

for i in range(len(records)):
    r = records[i]

    expected = hashlib.sha256(
        (r["evidence"] + str(r["time"]) +
         r["previous_hash"]).encode()
    ).hexdigest()

    if expected != r["hash"]:
        valid = False

    if i > 0:
        if r["previous_hash"] != records[i-1]["hash"]:
            valid = False
        if r["time"] < records[i-1]["time"]:
            valid = False

if valid:
    print("Chain of Custody: VALID")
else:
    print("Chain of Custody: INVALID")
```

**26. Static and Dynamic Malware Analysis**

```python
import hashlib

# Simulated malware sample
sample = b"CreateFile VirtualAlloc malware"

# Static analysis
file_hash = hashlib.sha256(sample).hexdigest()

print("Static Analysis")
print("SHA-256:", file_hash)

apis = ["CreateFile", "VirtualAlloc", "DeleteFile"]

for api in apis:
    if api.encode() in sample:
        print("Suspicious API:", api)

# Dynamic analysis (simulated behavior)
behavior = ["File created", "Registry modified", "Network connection"]

print("\nDynamic Analysis")

for action in behavior:
    print("Behavior:", action)
```

**28. DoS Attack Detection**

```python
logs = [
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.20"
]

limit = 3
count = {}

for ip in logs:
    count[ip] = count.get(ip, 0) + 1

for ip, total in count.items():
    print(ip, "Requests:", total)

    if total > limit:
        print("Possible DoS Attack Detected!")
```

**31. Packet-Filtering Firewall Rule Engine**

```python
rules = [
    ("192.168.1.10", "ALLOW"),
    ("192.168.1.20", "BLOCK")
]

ips = [
    "192.168.1.10",
    "192.168.1.20",
    "192.168.1.30"
]

for ip in ips:
    result = "DENY"

    for address, action in rules:
        if ip == address:
            result = action
            break

    print(ip, ":", result)
```
