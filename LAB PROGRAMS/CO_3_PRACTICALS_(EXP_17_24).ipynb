**17. ISO/IEC 27001 Annex A Control Domain Classification**

```python
def classify_control(description):
    text = description.lower()

    if "password" in text or "access" in text:
        return "Organizational / Access Control"
    elif "firewall" in text or "network" in text:
        return "Technological Controls"
    elif "building" in text or "physical" in text:
        return "Physical Controls"
    elif "employee" in text or "training" in text:
        return "People Controls"
    else:
        return "Control Domain Not Identified"


description = input("Enter security control description: ")

print("Control Domain:", classify_control(description))
```

**18. Malware Detection Using SHA-256 Hash**

```python
import hashlib
from google.colab import files

# Known malicious SHA-256 hash
malicious_hashes = [
    hashlib.sha256(b"malware").hexdigest()
]

# Upload file
uploaded = files.upload()

for filename in uploaded:
    file_data = uploaded[filename]

    # Calculate SHA-256 hash
    file_hash = hashlib.sha256(file_data).hexdigest()

    print("File Name:", filename)
    print("SHA-256 Hash:", file_hash)

    # Check malware
    if file_hash in malicious_hashes:
        print("Result: Malware Detected!")
    else:
        print("Result: File is Safe")
```

**19. Splunk Search Filter and Stats Count Simulation**

```python
events = [
    {"user": "admin", "status": "failed"},
    {"user": "john", "status": "success"},
    {"user": "admin", "status": "failed"},
    {"user": "john", "status": "failed"},
    {"user": "admin", "status": "success"}
]

# Search filter
print("Failed Login Events:")
for event in events:
    if event["status"] == "failed":
        print(event)

# Stats count by user
print("\nLogin Count by User:")
count = {}

for event in events:
    user = event["user"]
    count[user] = count.get(user, 0) + 1

for user, total in count.items():
    print(user, ":", total)
```

**22. Brute-Force Login Detection**

```python
user = input("Enter username to check: ")

logs = [
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "success"),
    ("john", "failed")
]

failed = 0
detected = False

for username, status in logs:
    if username == user:
        if status == "failed":
            failed += 1

        elif status == "success":
            if failed >= 3:
                detected = True
            failed = 0

if detected:
    print("Brute-force attack detected!")
else:
    print("No brute-force pattern detected")
```

**23. Intrusion Prevention System (IPS) with Whitelist**

```python
whitelist = ["192.168.1.10"]

blocked_signatures = ["malware", "attack", "sql injection"]

traffic = [
    ("192.168.1.10", "normal traffic"),
    ("192.168.1.20", "malware detected"),
    ("192.168.1.30", "normal traffic"),
    ("192.168.1.40", "sql injection detected")
]

for ip, data in traffic:

    if ip in whitelist:
        print(ip, "- Allowed (Trusted Source)")

    elif any(word in data.lower() for word in blocked_signatures):
        print(ip, "- Blocked (Threat Detected)")

    else:
        print(ip, "- Allowed")
```

**24. Security Control Baseline Checker**

```python
baseline = {
    "password_length": 8,
    "firewall": 1,
    "antivirus": 1,
    "backup": 1
}

current = {
    "password_length": 6,
    "firewall": 1,
    "antivirus": 0,
    "backup": 1
}

print("Security Control Report:")

for control, required in baseline.items():

    actual = current[control]

    if actual >= required:
        print(control, "- PASS")
    else:
        print(control, "- FAIL")
```
