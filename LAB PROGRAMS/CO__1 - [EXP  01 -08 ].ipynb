**1. Identify the Threat-Intelligence Application Area**

```python
def identify_area(scenario):
    scenario = scenario.lower()

    areas = {
        "Banking Fraud Prevention": [
            "bank", "banking", "payment fraud", "credit card",
            "financial transaction"
        ],
        "Healthcare Data Protection": [
            "hospital", "healthcare", "patient", "medical records",
            "health data"
        ],
        "E-commerce Fraud Prevention": [
            "e-commerce", "online shopping", "online store",
            "retail", "fake orders"
        ],
        "Cloud Security": [
            "cloud", "aws", "azure", "cloud storage"
        ]
    }

    for area, keywords in areas.items():
        if any(word in scenario for word in keywords):
            return area

    return "General Cybersecurity Threat Intelligence"


scenario = input("Enter a business scenario: ")
print("Relevant Application Area:", identify_area(scenario))
```

**2. Simple Vulnerability Assessment Using a Port Scanner**

```python
import socket

target = input("Enter target IP or hostname (use 127.0.0.1 for localhost): ")

# Scan only systems you own or have permission to test.
ports = [21, 22, 23, 25, 53, 80, 110, 443, 3306, 8080]

print("\nScanning:", target)

try:
    ip = socket.gethostbyname(target)

    for port in ports:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(0.5)

        result = sock.connect_ex((ip, port))

        if result == 0:
            print(f"Port {port}: OPEN")
        else:
            print(f"Port {port}: CLOSED or FILTERED")

        sock.close()

except socket.gaierror:
    print("Invalid hostname or unable to resolve host.")
```

**3. Security Policy Governance Checker**

```python
required_elements = {
    "Scope": ["scope", "applicability"],
    "Roles and Responsibilities": [
        "roles and responsibilities", "responsibilities"
    ],
    "Enforcement": ["enforcement", "disciplinary action"],
    "Review Cycle": ["review cycle", "reviewed annually", "review frequency"],
    "Approval Authority": ["approval authority", "approved by"]
}

policy = input("Enter your draft security policy:\n").lower()

missing = []

for element, keywords in required_elements.items():
    if not any(keyword in policy for keyword in keywords):
        missing.append(element)

print("\n--- Security Policy Assessment ---")

if missing:
    print("Missing governance elements:")
    for element in missing:
        print("-", element)
else:
    print("All required governance elements appear to be present.")

print("\nAssessment completed.")
```

**4. Demonstrate VPN Tunnel Confidentiality Using a Symmetric Stream Cipher**

```python
def xor_cipher(data, key):
    return bytes(
        byte ^ key[i % len(key)]
        for i, byte in enumerate(data)
    )


message = input("Enter message to send: ").encode()
key = input("Enter shared key: ").encode()

if not key:
    print("Error: Key cannot be empty.")
else:
    encrypted = xor_cipher(message, key)

    print("\nOriginal Message:", message.decode())
    print("Encrypted Data (hex):", encrypted.hex())

    # Simulate an attacker capturing the encrypted data
    captured_data = encrypted
    print("Captured Data (hex):", captured_data.hex())

    # Receiver decrypts using the same key
    decrypted = xor_cipher(captured_data, key)

    print("Decrypted Message:", decrypted.decode())
```

**5. Classify Threat-Intelligence Reports as Strategic, Tactical, or Operational**

```python
def classify_report(description, audience):
    text = (description + " " + audience).lower()

    categories = {
        "Strategic": [
            "business risk", "executives", "management",
            "long-term", "business impact", "risk assessment"
        ],
        "Tactical": [
            "attack techniques", "ttps", "security analysts",
            "attacker methods", "mitre", "attack patterns"
        ],
        "Operational": [
            "incident response", "soc team", "ip address",
            "malicious domain", "immediate action", "ioc",
            "active attack"
        ]
    }

    scores = {
        category: sum(keyword in text for keyword in keywords)
        for category, keywords in categories.items()
    }

    highest = max(scores.values())

    if highest == 0:
        return "Unclassified"

    winners = [
        category for category, score in scores.items()
        if score == highest
    ]

    if len(winners) > 1:
        return "Ambiguous: " + ", ".join(winners)

    return winners[0]


description = input("Enter report description: ")
audience = input("Enter intended audience: ")

print("Report Classification:",
      classify_report(description, audience))
```

**6. Phishing URL Detection Using Common Indicators**

```python
import re
from urllib.parse import urlparse

def detect_phishing(url):
    if not url.startswith(("http://", "https://")):
        url = "http://" + url

    parsed = urlparse(url)
    host = parsed.hostname or ""
    reasons = []

    suspicious_words = [
        "login", "verify", "update", "secure",
        "account", "free", "confirm"
    ]

    if parsed.scheme == "http":
        reasons.append("Uses unencrypted HTTP")

    if any(word in url.lower() for word in suspicious_words):
        reasons.append("Contains suspicious keywords")

    try:
        socket.inet_aton(host)
        reasons.append("Uses an IP address instead of a domain name")
    except OSError:
        pass

    if host.count(".") >= 3:
        reasons.append("Contains many subdomains")

    if "@" in parsed.netloc:
        reasons.append("Contains an @ symbol in the network location")

    if reasons:
        print("\nPotentially suspicious URL")
        for reason in reasons:
            print("-", reason)
    else:
        print("\nNo checked phishing indicators found.")

import socket

url = input("Enter a URL to check: ")
detect_phishing(url)
```

**7. Simulate the Three ELK Stack Roles Using Python**

```python
import re
from collections import Counter

# Raw logs: timestamp, level, and message
raw_logs = [
    "2026-10-16 10:00:00 INFO User logged in",
    "2026-10-16 10:01:00 ERROR Login failed",
    "2026-10-16 10:02:00 WARNING Multiple login attempts",
    "2026-10-16 10:03:00 ERROR Access denied",
    "2026-10-16 10:04:00 INFO Session closed"
]

# 1. Logstash-style parsing
events = []

for line in raw_logs:
    match = re.match(
        r"(\S+ \S+) (\w+) (.*)", line
    )

    if match:
        event = {
            "timestamp": match.group(1),
            "level": match.group(2),
            "message": match.group(3)
        }
        events.append(event)

print("1. Parsed Events:")
for event in events:
    print(event)

# 2. Elasticsearch-style indexing
index = {}

for i, event in enumerate(events, start=1):
    index[i] = event

print("\n2. Indexed Events:", len(index))

# 3. Kibana-style aggregation
counts = Counter(
    event["level"] for event in index.values()
)

print("\n3. Log Dashboard Summary:")
for level, count in sorted(counts.items()):
    print(f"{level}: {count}")
```

**8. Demonstrate Network Segmentation to Block Lateral Movement**

```python
# Define network segments
segments = {
    "Guest": ["192.168.1.10"],
    "Employee": ["192.168.2.10"],
    "Sensitive": ["192.168.3.10"]
}

# Define permitted communication paths
access_policy = {
    ("Guest", "Employee"): False,
    ("Guest", "Sensitive"): False,
    ("Employee", "Guest"): False,
    ("Employee", "Sensitive"): False,
    ("Sensitive", "Employee"): True,
    ("Sensitive", "Guest"): False
}

def check_access(source, destination):
    if source == destination:
        return True

    return access_policy.get(
        (source, destination), False
    )


source = "Guest"
destination = "Sensitive"

print("Network Segmentation Simulation")
print("Compromised Segment:", source)
print("Target Segment:", destination)

if check_access(source, destination):
    print("ACCESS GRANTED")
    print("Communication is permitted.")
else:
    print("ACCESS DENIED")
    print("Segmentation blocks lateral movement.")

print("\nAdditional Access Tests:")

tests = [
    ("Guest", "Employee"),
    ("Guest", "Sensitive"),
    ("Employee", "Sensitive"),
    ("Sensitive", "Employee")
]

for src, dst in tests:
    status = (
        "ALLOWED" if check_access(src, dst)
        else "BLOCKED"
    )
    print(f"{src} -> {dst}: {status}")
```
